## Occupational Compensation Analysis

### Business Questions 

This phase examines Canadian occupational wage data to evaluate compensation levels, market competitiveness, and wage trends, with a particular focus on Nova Scotia relative to the Canadian labour market.

The analysis addresses the following business questions:

#### Compensation Benchmarking
1. What are the median hourly wages across major occupational groups?
2. How do Nova Scotia median hourly wages compare with corresponding Canadian benchmarks?
3. Which occupational groups in Nova Scotia are below, near, or above the Canadian market benchmark?
4. Which occupational groups show the largest positive and negative market wage gaps?

#### Market Competitiveness
5. Which Nova Scotia occupational groups may warrant further compensation review based on their market position?
6. How competitive are professional occupations in finance, business, government services, health, education, sciences, engineering, and other major occupational groups?
7. Are compensation differences consistent across occupational groups, or do market positions vary substantially?

#### Wage Trend Analysis
8. How have median hourly wages changed in Nova Scotia between 2015 and 2025?
9. How have Canadian median hourly wages changed over the same period?
10. Which occupational groups experienced the strongest and weakest wage growth in Nova Scotia?
11. Did Nova Scotia wages grow faster or slower than corresponding Canadian wages?

#### Market Gap Trend Analysis
12. Has Nova Scotia's compensation position relative to Canada improved or deteriorated between 2015 and 2025?
13. Which occupational groups experienced the greatest improvement in relative market position?
14. Which occupational groups experienced the greatest deterioration?
15. Are current compensation gaps persistent, narrowing, or widening over time?

#### Compensation Decision Support
16. Which occupational groups appear most exposed to potential market competitiveness concerns?
17. Which occupational groups have moved from below-market to near-market or above-market positions?
18. Which unusual compensation movements require additional investigation before conclusions are drawn?
19. How can external labour-market data support evidence-based compensation review and workforce planning?

> **Analytical Note:** Market comparisons in this analysis are indicators of external labour-market positioning and are not direct recommendations for pay adjustments. Compensation decisions would require additional information such as job evaluation, internal pay structures, employee experience, benefits, pensions, collective agreements, and other components of total compensation.

```python
import pandas as pd
import numpy as np

occupation_3digit = pd.read_csv(
    "../data/processed/occupation_benchmark_clean.csv",
    dtype={"noc_code": str}
)
```

#### Nova Scotia vs. Canada Compensation Benchmark

To evaluate compensation competitiveness, median hourly wages for occupational groups in Nova Scotia are compared with corresponding Canadian wage benchmarks.

The most recent available year, 2025, is used for the initial market comparison.

A market gap is calculated as:

**Market Gap (%) = (Nova Scotia Wage − Canada Wage) / Canada Wage × 100**

A negative market gap indicates that the Nova Scotia median wage is below the corresponding national benchmark, while a positive value indicates that it is above the national benchmark.

```python
#create the 2025 data:
latest = occupation_3digit[
    occupation_3digit["year"] == 2025
].copy()
```

```python
#Canada:
canada_2025 = latest[
    latest["geography"] == "Canada"
][
    ["noc_code", "Occupation", "median_hourly_wage"]
].rename(
    columns={"median_hourly_wage": "canada_wage"}
)
```

```python
#Nova Scotia:
ns_2025 = latest[
    latest["geography"] == "Nova Scotia"
][
    ["noc_code", "Occupation", "median_hourly_wage"]
].rename(
    columns={"median_hourly_wage": "nova_scotia_wage"}
)
```

```python
#now merge Canada and Nova Scotia:
ns_vs_canada = ns_2025.merge(
    canada_2025,
    on=["noc_code", "Occupation"],
    how="inner"
)
```

#### Market Gap Analysis

```python
#Calculate the market gap:
ns_vs_canada["market_gap_pct"] = (
    (ns_vs_canada["nova_scotia_wage"] -
     ns_vs_canada["canada_wage"])
    / ns_vs_canada["canada_wage"]
    * 100
)

#Round it:

ns_vs_canada[
    ["nova_scotia_wage", "canada_wage", "market_gap_pct"]
] = ns_vs_canada[
    ["nova_scotia_wage", "canada_wage", "market_gap_pct"]
].round(2)
```

```python
ns_vs_canada.sort_values("market_gap_pct")
```

#### Key Findings — Nova Scotia vs. Canada

The 2025 compensation benchmark reveals notable differences between median hourly wages in Nova Scotia and national wage levels across the selected occupational groups.

- **10 of the 14 occupational groups** analyzed have median hourly wages below the Canadian benchmark.
- **Professional occupations in finance (NOC 111)** show the largest negative market gap, with a Nova Scotia median hourly wage of **$36.20** compared with **$43.00 nationally**, representing a **15.81% gap**.
- Professional occupations in **social and community services (-9.85%)**, **education services (-9.55%)**, and **natural sciences (-8.96%)** also show relatively large negative gaps.
- Nova Scotia's median wages for **professional occupations in business** are relatively close to the national benchmark, with a gap of only **-1.86%**.
- Four occupational groups have Nova Scotia wages at or above the Canadian benchmark.
- **Professional occupations in law (NOC 411)** show the largest positive difference, with a Nova Scotia median hourly wage of **$74.52**, approximately **19.23% above** the Canadian benchmark.
- **Nursing and allied health professionals** are approximately **4.35% above** the national benchmark, while health treating and consultation professionals are approximately **2.94% above**.

#### Interpretation Considerations

The market gaps should be interpreted as indicators of broader labour-market differences rather than direct recommendations for pay adjustments.

Compensation decisions would require additional information, including job responsibilities, internal job evaluation, organizational pay structures, employee experience, benefits, pensions, collective agreements, and other components of total compensation.

The benchmark is therefore most useful for identifying occupational groups that may warrant further compensation review.

#### Market Competitiveness Classification

To support interpretation of the market benchmark, occupational groups are classified according to their percentage difference from the Canadian median wage.

- **Below Market:** more than 5% below the Canadian benchmark
- **Near Market:** within ±5% of the Canadian benchmark
- **Above Market:** more than 5% above the Canadian benchmark

These classifications are analytical indicators and should not be interpreted as direct recommendations for compensation adjustments.

```python
def classify_market(gap):
    if gap < -5:
        return "Below Market"
    elif gap > 5:
        return "Above Market"
    else:
        return "Near Market"

ns_vs_canada["market_position"] = (
    ns_vs_canada["market_gap_pct"].apply(classify_market)
)

ns_vs_canada.sort_values("market_gap_pct")
```

```python
ns_vs_canada["market_position"].value_counts()
```

```python
import matplotlib.pyplot as plt

plot_data = ns_vs_canada.sort_values("market_gap_pct")

plt.figure(figsize=(10, 7))

plt.barh(
    plot_data["Occupation"],
    plot_data["market_gap_pct"]
)

plt.axvline(0, linewidth=1)

plt.xlabel("Market Gap (%)")
plt.ylabel("Occupational Group")
plt.title(
    "Nova Scotia Median Hourly Wage Gap vs. Canada — 2025"
)

plt.tight_layout()
plt.show()
```

#### Interpretation

Most occupational groups analyzed have median hourly wages below the Canadian benchmark. The largest negative market gap is observed in professional finance occupations, while professional occupations in law show a substantial positive difference.

The results suggest that compensation competitiveness varies considerably across occupational groups, highlighting the importance of occupation-specific market benchmarking rather than relying solely on overall wage averages.

#### Compensation Market Gap Trends — 2015–2025

A single-year market comparison provides a snapshot of compensation competitiveness but does not indicate whether wage differences are improving or worsening over time.

This analysis compares Nova Scotia median hourly wages with corresponding Canadian benchmarks from 2015 to 2025.

The longitudinal analysis helps identify occupational groups where compensation gaps are:

- Widening over time
- Narrowing over time
- Relatively stable
- Moving above or below the national benchmark

```python
#create the Canada dataset for all years:
canada_trend = occupation_3digit[
    occupation_3digit["geography"] == "Canada"
][
    ["year", "noc_code", "Occupation", "median_hourly_wage"]
].rename(
    columns={"median_hourly_wage": "canada_wage"}
)

canada_trend.head()
```

```python
#for Nova Scotia:
ns_trend = occupation_3digit[
    occupation_3digit["geography"] == "Nova Scotia"
][
    ["year", "noc_code", "Occupation", "median_hourly_wage"]
].rename(
    columns={"median_hourly_wage": "nova_scotia_wage"}
)

ns_trend.head()
```

```python
#Merge them
market_trend = ns_trend.merge(
    canada_trend,
    on=["year", "noc_code", "Occupation"],
    how="inner"
)

market_trend.shape
```

```python
#calculate the percentage gap:
market_trend["market_gap_pct"] = (
    (market_trend["nova_scotia_wage"] -
     market_trend["canada_wage"])
    / market_trend["canada_wage"]
    * 100
)

market_trend["market_gap_pct"] = (
    market_trend["market_gap_pct"].round(2)
)
```

```python
market_trend.head()
```

```python
market_trend.isnull().sum()
```

```python
market_trend.groupby("year")[
    "market_gap_pct"
].count()
```

Finance had the largest negative gap in 2025, so let's see whether that gap is improving or worsening.

```python
finance_trend = market_trend[
    market_trend["noc_code"] == "111"
][
    [
        "year",
        "nova_scotia_wage",
        "canada_wage",
        "market_gap_pct"
    ]
]

finance_trend
```

```python
plt.figure(figsize=(9, 5))

plt.plot(
    finance_trend["year"],
    finance_trend["market_gap_pct"],
    marker="o"
)

plt.axhline(0, linewidth=1)

plt.xlabel("Year")
plt.ylabel("Market Gap (%)")
plt.title(
    "Nova Scotia vs. Canada Wage Gap\n"
    "Professional Occupations in Finance (NOC 111)"
)

plt.xticks(finance_trend["year"], rotation=45)

plt.tight_layout()
plt.show()
```

Now let's calculate how the market position changed over the decade.

```python
#extract 2015:
gap_2015 = market_trend[
    market_trend["year"] == 2015
][
    ["noc_code", "Occupation", "market_gap_pct"]
].rename(
    columns={"market_gap_pct": "gap_2015"}
)
```

```python
#2025:
gap_2025 = market_trend[
    market_trend["year"] == 2025
][
    ["noc_code", "Occupation", "market_gap_pct"]
].rename(
    columns={"market_gap_pct": "gap_2025"}
)
```

```python
#Merge them
gap_change = gap_2015.merge(
    gap_2025,
    on=["noc_code", "Occupation"]
)
```

```python
gap_change["gap_change"] = (
    gap_change["gap_2025"] -
    gap_change["gap_2015"]
).round(2)
```

```python
gap_change.sort_values("gap_change")
```

#### Key Findings — Change in Market Position

The comparison between 2015 and 2025 shows that Nova Scotia's relative compensation position improved for several occupational groups, although the direction and magnitude of change varied considerably.

- **9 of the 14 occupational groups** improved their relative position compared with the Canadian benchmark.
- Professional occupations in business improved substantially, moving from **17.65% below** the Canadian benchmark in 2015 to only **1.86% below** in 2025.
- Nursing and allied health professionals moved from **3.69% below** the national benchmark to **4.35% above** it.
- Health treating and consultation services professionals moved from **6.23% below** the national benchmark to **2.94% above** it.
- Paraprofessional occupations in legal, social, community, education, and related services improved from **12.50% below** the Canadian benchmark to approximately market level in 2025.
- Professional occupations in finance improved slightly, but remained substantially below the national benchmark at **15.81% below** in 2025.
- Professional occupations in social and community services experienced the largest deterioration, moving from **4.17% above** the Canadian benchmark in 2015 to **9.85% below** in 2025.

Professional occupations in law show an unusually large change and require additional investigation before drawing conclusions.

#### Calculate actual wage growth

There's another important distinction.

Market gap change ≠ wage growth.

For example, Nova Scotia wages could increase significantly but still fall behind Canada if Canadian wages increased faster.

Let's calculate actual wage growth between 2015 and 2025.


```python
wages_2015 = market_trend[
    market_trend["year"] == 2015
][
    [
        "noc_code",
        "Occupation",
        "nova_scotia_wage",
        "canada_wage"
    ]
].rename(columns={
    "nova_scotia_wage": "ns_wage_2015",
    "canada_wage": "canada_wage_2015"
})
```

```python
wages_2025 = market_trend[
    market_trend["year"] == 2025
][
    [
        "noc_code",
        "Occupation",
        "nova_scotia_wage",
        "canada_wage"
    ]
].rename(columns={
    "nova_scotia_wage": "ns_wage_2025",
    "canada_wage": "canada_wage_2025"
})
```

```python
wage_growth = wages_2015.merge(
    wages_2025,
    on=["noc_code", "Occupation"]
)
```

```python
#Calculate growth:
wage_growth["ns_growth_pct"] = (
    (
        wage_growth["ns_wage_2025"] -
        wage_growth["ns_wage_2015"]
    )
    / wage_growth["ns_wage_2015"]
    * 100
).round(2)

wage_growth["canada_growth_pct"] = (
    (
        wage_growth["canada_wage_2025"] -
        wage_growth["canada_wage_2015"]
    )
    / wage_growth["canada_wage_2015"]
    * 100
).round(2)
```

```python
wage_growth["growth_difference"] = (
    wage_growth["ns_growth_pct"] -
    wage_growth["canada_growth_pct"]
).round(2)

wage_growth[
    [
        "noc_code",
        "Occupation",
        "ns_growth_pct",
        "canada_growth_pct",
        "growth_difference"
    ]
].sort_values("growth_difference")
```

#### Why this is an important compensation metric

Now we can distinguish between two questions:

##### Market position

How much higher/lower is Nova Scotia's wage than Canada's?

##### Market movement

Are Nova Scotia wages growing faster or slower than the Canadian benchmark?

### Key Findings — Wage Growth, 2015–2025

The wage growth analysis provides additional context for changes in Nova Scotia's relative market position.

- Median hourly wages increased across all 14 occupational groups analyzed in Nova Scotia between 2015 and 2025.
- **Professional occupations in social and community services (NOC 413)** experienced the weakest relative growth. Nova Scotia wages increased by **15.39%**, compared with **33.33% nationally**, a difference of approximately **17.94 percentage points**.
- **Professional occupations in business (NOC 112)** experienced strong wage growth in Nova Scotia of **50.97%**, compared with **26.67% nationally**, substantially narrowing the compensation gap with the Canadian benchmark.
- **Nursing and allied health professionals (NOC 313)** recorded wage growth of **33.33%** in Nova Scotia compared with **23.06% nationally**, contributing to Nova Scotia moving above the national benchmark by 2025.
- **Health treating and consultation services professionals (NOC 311)** also experienced stronger growth in Nova Scotia, at **35.17%**, compared with **23.13% nationally**.
- Finance occupations experienced slightly stronger wage growth in Nova Scotia than nationally; however, their 2025 median hourly wage remained substantially below the Canadian benchmark.
- Professional occupations in law show an unusually large increase and should be investigated further before the result is interpreted as a sustained market trend.

Overall, the results demonstrate that wage growth alone does not determine market competitiveness. Both the current market position and changes in compensation relative to external benchmarks should be considered when assessing compensation trends.

> **Note:** Wage growth rates are based on current-dollar wages and are not adjusted for inflation. Therefore, they represent nominal wage growth rather than changes in real purchasing power.

#### summary table

```python
occupation_scorecard = ns_vs_canada.merge(
    wage_growth[
        [
            "noc_code",
            "ns_growth_pct",
            "canada_growth_pct",
            "growth_difference"
        ]
    ],
    on="noc_code",
    how="left"
)

occupation_scorecard = occupation_scorecard[
    [
        "noc_code",
        "Occupation",
        "nova_scotia_wage",
        "canada_wage",
        "market_gap_pct",
        "market_position",
        "ns_growth_pct",
        "canada_growth_pct",
        "growth_difference"
    ]
]

occupation_scorecard.sort_values("market_gap_pct")
```
