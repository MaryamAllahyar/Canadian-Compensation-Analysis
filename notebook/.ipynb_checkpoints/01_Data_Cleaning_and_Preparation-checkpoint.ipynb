#### Identify Key Variables and Categories

The Statistics Canada datasets contain several administrative and metadata fields that are not required for the compensation analysis.

The analysis focuses on variables related to time, geography, wage measures, employment type, occupation or industry, demographic characteristics, and reported wage values.

```python
occupation_columns = [
    "REF_DATE",
    "GEO",
    "Wages",
    "Type of work",
    "National Occupational Classification (NOC)",
    "Gender",
    "Age group",
    "UOM",
    "VALUE"
]

industry_columns = [
    "REF_DATE",
    "GEO",
    "Wages",
    "Type of work",
    "North American Industry Classification System (NAICS)",
    "Gender",
    "Age group",
    "UOM",
    "VALUE"
]

#occupation dataset:
occupation_categories = {
    "REF_DATE": set(),
    "GEO": set(),
    "Wages": set(),
    "Type of work": set(),
    "Gender": set(),
    "Age group": set(),
    "UOM": set()
}

for chunk in pd.read_csv(
    occupation_path,
    usecols=occupation_columns,
    chunksize=200000
):
    for col in occupation_categories:
        occupation_categories[col].update(
            chunk[col].dropna().unique()
        )

for col, values in occupation_categories.items():
    print(f"\n{col}:")
    print(sorted(values))
```

#### Define the Primary Compensation Benchmark

To ensure meaningful and consistent compensation comparisons, a standardized benchmark population was defined.

The primary benchmark focuses on:

- Full-time employees
- Total gender population
- Employees aged 15 years and over
- Median hourly wage rates
- Canadian provinces and the national benchmark
- 2015–2025 analysis period

Median hourly wages were selected as the primary compensation measure because the median is less sensitive to unusually high or low wage values and provides a useful representation of typical market compensation.

Additional datasets will be created separately for demographic and gender-based compensation analysis.

```python
benchmark_chunks = []

for chunk in pd.read_csv(
    occupation_path,
    usecols=occupation_columns,
    chunksize=200000
):

    filtered = chunk[
        (chunk["REF_DATE"] >= 2015) &
        (chunk["Wages"] == "Median hourly wage rate") &
        (chunk["Type of work"] == "Full-time employees") &
        (chunk["Gender"] == "Total - Gender") &
        (chunk["Age group"] == "15 years and over") &
        (chunk["UOM"] == "Current dollars")
    ]

    benchmark_chunks.append(filtered)

occupation_benchmark = pd.concat(
    benchmark_chunks,
    ignore_index=True
)

occupation_benchmark.shape
```

#### Standardize Occupational Classification Level

The occupation dataset contains multiple levels of the National Occupational Classification (NOC) hierarchy, ranging from broad occupational categories to more detailed occupational groups.

Comparing wage levels across different hierarchy levels could produce misleading results. Therefore, occupational codes are extracted from the occupation descriptions so that comparable classification levels can be identified and used for compensation benchmarking.

```python
#Extract the NOC code
occupation_benchmark["NOC_Code"] = (
    occupation_benchmark["National Occupational Classification (NOC)"]
    .str.extract(r"\[([^\]]+)\]")
)

occupation_benchmark[
    ["National Occupational Classification (NOC)", "NOC_Code"]
].head(20)
```

```python
#Determine the code structure
occupation_benchmark["NOC_Code"].value_counts().sort_index()
```

```python
#rather than manually looking through everything, let's classify the codes.
def classify_noc(code):
    if pd.isna(code):
        return "Unknown"
    elif code.isdigit():
        return f"{len(code)}-digit"
    else:
        return "Aggregate/Multiple"

occupation_benchmark["NOC_Level"] = (
    occupation_benchmark["NOC_Code"].apply(classify_noc)
)

occupation_benchmark["NOC_Level"].value_counts()
```

```python
#check how many occupational categories exist at each level
occupation_benchmark.groupby("NOC_Level")[
    "National Occupational Classification (NOC)"
].nunique()
```

```python
#inspect the 3-digit occupations
three_digit_occupations = (
    occupation_benchmark.loc[
        occupation_benchmark["NOC_Level"] == "3-digit",
        ["NOC_Code", "National Occupational Classification (NOC)"]
    ]
    .drop_duplicates()
    .sort_values("NOC_Code")
)

three_digit_occupations
```

#### Create a Comparable Occupational Benchmark

The source data contains multiple levels of the National Occupational Classification (NOC) hierarchy. To improve comparability, the primary occupational benchmark is restricted to three-digit NOC groups.

This produces a consistent set of occupational groups that can be compared across provinces and over time without mixing broad occupational categories with more detailed classifications.

The resulting benchmark will be used to evaluate:

- Compensation levels across occupational groups
- Provincial differences in compensation
- Nova Scotia wages relative to the Canadian benchmark
- Long-term wage trends
- Market competitiveness

```python
#create the dataset:
occupation_3digit = occupation_benchmark[
    occupation_benchmark["NOC_Level"] == "3-digit"
].copy()

occupation_3digit.shape
```

```python
#reate a new column:
occupation_3digit["Occupation"] = (
    occupation_3digit["National Occupational Classification (NOC)"]
    .str.replace(r"\s*\[[^\]]+\]$", "", regex=True)
    .str.strip()
)
```

```python
occupation_3digit[
    ["NOC_Code", "Occupation"]
].drop_duplicates().sort_values("NOC_Code")
```

## Standardize Column Names

Column names are standardized to improve readability and maintain consistency across Python, SQL, and Power BI.

```python
occupation_3digit = occupation_3digit.rename(columns={
    "REF_DATE": "year",
    "GEO": "geography",
    "Wages": "wage_measure",
    "Type of work": "work_type",
    "Gender": "gender",
    "Age group": "age_group",
    "UOM": "unit",
    "VALUE": "median_hourly_wage",
    "NOC_Code": "noc_code",
    "NOC_Level": "noc_level"
})
```

```python
#Now we no longer need the original long NOC column:
occupation_3digit = occupation_3digit.drop(
    columns=["National Occupational Classification (NOC)"]
)
```
